```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


obj = Dog()

obj.eat()
obj.bark()
```

```python
class Grandfather:
    def house(self):
        print("Grandfather has a house")


class Father(Grandfather):
    def car(self):
        print("Father has a car")


class Son(Father):
    def bike(self):
        print("Son has a bike")


obj = Son()

obj.house()
obj.car()
obj.bike()
```

```python
class Father:
    def money(self):
        print("Father has money")


class Mother:
    def gold(self):
        print("Mother has gold")


class Child(Father, Mother):
    def education(self):
        print("Child has education")


obj = Child()

obj.money()
obj.gold()
obj.education()
```

```python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


class Cat(Animal):
    def meow(self):
        print("Cat is meowing")


dog = Dog()
dog.eat()
dog.bark()

cat = Cat()
cat.eat()
cat.meow()
```

```python
class Parent:
    def __init__(self):
        print("Parent Constructor")


class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Constructor")


obj = Child()
```

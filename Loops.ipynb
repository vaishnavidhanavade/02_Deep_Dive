```python
#Nested Loop
col =["Red","Yellow","Green"]
fruits =["Apple","Bannana","Grapes"]
for i in col:
    for j in fruits:
        print(i,j)
```

```python
col_01=["Red","Yellow","LightGreen"]
col_02=["White","Blue","LightGreen"]
for i in col_01:
    for j in col_02:
        if i==j:
            print("Matched=",i,j)
```

```python
for j in range(1,7):
    for i in range(1,j,-1):
        print(" ",end=" ")
    for i in range(1,j+1):
        print("*",end="")
    print()
```

```python
for i in range(1,7):
    for j in range(6-i):
        print("_",end=" ")
    for k in range(1,i+1):
        print("*",end="")
    print()
```

```python
n = 5

# Upper part
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()

# Lower part
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")
    
    for j in range(2 * i - 1):
        print("*", end="")
    
    print()
```

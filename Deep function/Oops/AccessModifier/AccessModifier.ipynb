```python
class student:
    def __init__(self):
        self.name = "vaishnavi"    #public
        self.age = 23              #Protected
        self.__marks = 90          #Private

    def display(self):
        print(self.name)           #public
        print(self.age)            #Protected
        print(self.__marks)        #Private

obj = student()
obj.display()

      
```

```python
# Outside the class
class student:
    def __init__(self):
        self.name = "vaishnavi"        #public
        self._age = 20                 #Protected
        self.__marks = 90              #Private
obj = student()

# Access outside the class
print(obj.name)         #public = Allowed
print(obj._age)         #Protected = Allowed
print(obj.__marks)      #Private = Error
```

```python
class Student:

    def display_public(self):
        print("Public Method")

    def _display_protected(self):
        print("Protected Method")

    def __display_private(self):
        print("Private Method")


class Teacher(Student):
    def show(self):
        self.display_public()       # Allowed
        self._display_protected()   # Allowed
        self.__display_private()    # Error


obj = Teacher()
obj.show()
```
